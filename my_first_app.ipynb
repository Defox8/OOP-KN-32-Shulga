from datetime import datetime

  Ми підключаємо стандартний модуль Python для роботи з датою та часом. Це позбавляє нас необхідності вручну прописувати поточний рік (код сам знатиме, який зараз рік).


  

  

  

input("...") та int(...)


  

  
Функція input() зупиняє виконання програми та очікує, поки користувач введе текст із клавіатури.

  
За замовчуванням input() повертає текст (str). Щоб відняти рік народження від поточного року, ми перетворюємо текст на ціле число за допомогою int().

  

  

  

  

Обчислення (age = current_year - birth_year)

  Звичайна арифметика: беремо поточний рік через datetime.now().year та віднімаємо введений рік.


  

  

  

Розгалуження (if / elif / else)

  Це «мозок» програми — вона приймає рішення на основі отриманих даних. Якщо умова age < 18 правдива, виконається перший блок; якщо ні — перевіряється наступна умова (elif), або спрацьовує гілка за замовчуванням (else).


  

  

  

f-рядки (f"Привіт, {name}...")

  Літера f перед лапками дозволяє зручно вставляти значення змінних безпосередньо у фігурні дужки {} всередині тексту.
